```python
import pymysql
import matplotlib.pyplot as plt
```

```python
myDB = pymysql.connect(host="localhost", user="root", password="", database="tec_info")
myCursor = myDB.cursor()
```

```python
myCursor.execute("SELECT Nombre, Promedio FROM Estudiantes ORDER BY Promedio LIMIT 5")
result = myCursor.fetchall
```

```python
Nombres = []
Calificaciones = []

for i in myCursor:
    Nombres.append(i[0])
    Calificaciones.append(i[1])

print("Nombre de los estudiantes = ", Nombres)
print("Califs. de los estudiantes = ", Calificaciones)
```

```python
plt.bar(Nombres, Calificaciones)
plt.xlabel("Nombres")
plt.ylabel("Calificaciones")
plt.title("Los alumnos de peor promedio")
plt.ylim(5.9, 6.2)
```
